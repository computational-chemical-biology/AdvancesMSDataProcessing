## 1. Introduction <a id="sec-intro"></a>Data quality assessment is an integral part of preparatory data analysis to ensure sound biological information retrieval. We present here the `MatrixQCvisUtils` package, which provides functions to create `SummarizedExperiment` objects.The `SummarizedExperiment` objects can be seamlessly used in the `MatrixQCvis` package, that provides shiny-based interactive visualization of data quality metrics at the per-sample and per-feature level.

## 3. Load files and convert to a `SummarizedExperiment` objectThe functions in `MatrixQCvisUtils` convert the output files of the following workflows into a `SummarizedExperiment`:- Biocrates/MetIDQ output in xlsx format,- MaxQuant output in xlsx or txt format, and- Spectronaut output in xlsx format.The functions `biocrates`, `maxQuant`, or `spectronaut` will create `MatrixQCvis::shinyQC` compatible `SummarizedExperiment` objects, with the following properties:- `rownames(se)` are set to the feature names derived from the respective cells in the input files,- `colnames(se)` are set to the sample names derived from the respective cells in the input files,- `colnames(se)`, `colnames(assay(se))` and `rownames(colData(se))` are identical.Objects belonging to the `SummarizedExperiment` class are containers for one or more assays, which are (numerical) matrices containing the quantitative, measured information of the experiment. The rows represent features of interest (e.g. transcripts, peptides, proteins, or metabolites) and the columns represent the samples. The `SummarizedExperiment` object stores also information on the features of interest (accessible by `rowData`) and information on the samples (accessible by `colData`). The name of samples and features will be accessed from `colnames(se)` and `rownames(se)`, respectively.

### 3.1 Conversion from Biocrates files

```r
file <- system.file("extdata", "biocrates_test_file.xlsx",     package = "MatrixQCvisUtils")se <- biocrates(file = file, sheet = 1)se
```

### 3.2 Conversion from Metaboscape files

```r
file <- system.file("extdata", "metaboscape_test_file.xlsx",     package = "MatrixQCvisUtils")se <- metaboscape(file = file, sheet = 1)se
```

### 3.3 Conversion from MaxQuant files

```r
file <- system.file("extdata", "maxquant_test_file.xlsx",         package = "MatrixQCvisUtils")se <- maxquant(file = file, type = "xlsx", intensity = "LFQ", sheet = 1)se
```

Additional ways of converting MaxQuant files into a `SummarizedExperiment` object.> Note: `eval = FALSE` in the vignette — the paths below are placeholders, replace them with your own files before running.

```r
## from txt files and iBAQ-normalized intensitiesfile <- "path/to/maxQuant/object.txt"maxQuant(file = file, intensity = "iBAQ", type = "txt")## from txt files and LFQ-normalized intensitiesfile <- "path/to/maxQuant/object.txt"maxQuant(file = file, intensity = "LFQ", type = "txt")## from xlsx files and iBAQ-normalized intensitiesfile <- "path/to/maxQuant/object.xslx"maxQuant(file = file, intensity = "iBAQ", type = "xlsx")## from xlsx files and LFQ-normalized intensities (see also above)file <- "path/to/maxQuant/object.xlsx"maxQuant(file = file, intensity = "LFQ", type = "xlsx")
```

### 3.4 Conversion from DIA-NN files

```r
file <- system.file("extdata", "diann_test_file.tsv",         package = "MatrixQCvisUtils")se <- diann(file = file)se
```

### 3.5 Conversion from Spectronaut files> Note: `eval = FALSE` in the vignette.

```r
file <- system.file("extdata", "spectronaut_test_file.xlsx",         package = "MatrixQCvisUtils")se <- spectronaut(file = file, sheetIntensities = 1, sheetAnnotation = 2)se
```
