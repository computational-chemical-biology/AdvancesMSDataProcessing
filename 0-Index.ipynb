# Advanced Mass Spectrometry Data Processing

Interactive notebooks for the **Advances in MS Data Processing** track of the
[5th IberoAmerican School on Advanced Mass Spectrometry](https://5iberoamerican.brmass.com/)
(Rio de Janeiro, September 28 - October 2, 2026).

This online book collects the hands-on notebooks used during the course. The material is
organized following the course program:

## Introduction

- PyOpenMS prerequisites

## SpectralDataPreprocessingTreatment

- Metabolomics trail: spectral preprocessing (Bacteria_Proc_Methods, pyOpenMS)
- Proteomics trail: peak processing and Python in Proteomics

## StatisticalValidationModelRobustness

- QC-MXP demo

## Cheminformatics AppliedMassSpectrometryStructuralAnalysis

- PubChem elucidation (ms-pred) and PyOpenMS identification

## MultivariateExploratoryAnalysisSupervisedModelsClassificationPrediction

- Univariate statistics and OmicVerse metabolomics tutorials

## PredictiveModelingBioactivityChemicalProperties

- FERMO and Chemprop

## DataIntegrationMassSpectrometry

- Multi-omics data integration

## EmergingComputationalPlatformsPipelines

- DreaMS beer profiler, AlphaPept, protein-ligand modelling

## How to use this book

- Read the static pages in your browser, or
- Run the notebooks interactively in JupyterLab (see the [README](README.md) for setup),
- Most notebooks can also be executed for free in Google Colab by clicking the badge at the top of each page.
